```python
#plot a circle

import matplotlib.pyplot as plt
import numpy as np

fig,ax = plt.subplots()

# fig= overall figure
# ax= the axes (where the plot will be drawn)

circle = plt.Circle((5,5),2,color='orange',fill=False,linewidth=2)
ax.add_artist(circle)
ax.set_xlim(0,10)
ax.set_ylim(0,10)
ax.set_aspect('equal')
plt.show()
```

```python
#plot a rectangle

fig,ax = plt.subplots()

rect=plt.Rectangle((5,5),4,2,fill=False,linewidth=2,color='orange')
ax.add_artist(rect)
ax.set_xlim(0,10)
ax.set_ylim(0,10)
ax.set_aspect('equal')
plt.show()
```

##Sympy Library

```python
#Sympy
#Define symbolic variables
import sympy as sp

x,y=sp.symbols('x y')

expr= x**2 + 2*x*y + y**2

print(expr) #print a plain text version

expr
```

```python
from sympy import *

x,y=sp.symbols('x y')
expr = (x+2)*(x+3)

#Expanded
expanded = expand(expr)
print(expanded)
expanded

```

```python
#Factored
factored = factor(expanded)
print(factored)
factored
```

```python
expr2 = 5*x**2 - 3*x**2

#Simplify

simplified= simplify(expr2)
# print(simplified)
simplified
```

```python
# Differentiation and Integration

x=sp.symbols('x')
f = sin(x)+x**2
integrate(f,x)
```

```python
#system of equations

x,y=sp.symbols('x y')
eq1= Eq(x**2-4,0)
solve(eq1,x)

```

```python
eq2= Eq(x+y,5)
eq3= Eq(2*x-y,1)

solve((eq2,eq3),(x,y))
```

```python
# Matrix Operations

A =Matrix([[2,3],[1,4]])
A
```

```python
print("Determinant:" ,A.det())
print("Inverse:" ,A.inv())
print("Transpose:" ,A.T)
A.T
```

```python
#Eigen values and eigen vectors

print("Eigen Values:", A.eigenvals())
print("Eigen Vectors:", A.eigenvects())

```

```python
#Matrix Multiplication

A =Matrix([[2,3],[1,4]])
B =Matrix([[5,3],[2,4]])

C= A*B

print(C)
C
```
