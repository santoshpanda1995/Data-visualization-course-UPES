```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load dataset

data = sns.load_dataset('tips')
```

```python
data.head()
```

```python
data
```

```python
#Show Available Datasets in seaborn


sns.get_dataset_names()
```

```python
#Scatterplot

sns.scatterplot(x="total_bill", y="total_bill",data=data)
```

```python
sns.scatterplot(x="total_bill", y="tip",data=data)
```

```python
#Line Plot

sns.lineplot(x="size", y="tip",data=data)
```

```python
#Boxplot

sns.boxplot(x="day",y="total_bill",data=data)
```

```python
#Violin plot
sns.violinplot(x="day",y="total_bill",data=data)
```

```python
#BarPlot

sns.barplot(x="day",y="total_bill",data=data)
```

```python
#BarPlot without errorbar

sns.barplot(x="day",y="total_bill",data=data, errorbar=None)
```

```python
#countplot

sns.countplot(x="day",data=data)
```

```python
#Swarmplot

sns.swarmplot(x="day",y="total_bill",data=data)
```

```python
#Histogram

sns.histplot(x="tip",data=data)
```

```python
#Heatmap

sns.heatmap(data.select_dtypes(include='number').corr(),annot=True,cmap="viridis")
```

```python
#PairPlot

sns.pairplot(data)
```
