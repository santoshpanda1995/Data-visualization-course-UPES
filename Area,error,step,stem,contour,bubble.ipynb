```python
#Area plot

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug"]
sales = [120, 180, 250, 300, 400, 450, 500, 600]

plt.fill_between(months,sales,color="skyblue",alpha=0.5,label="Sales Growth")
plt.plot(months,sales,color="blue", marker="o")

plt.title("Monthly Sales growth")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
#Error Plot
import numpy as np

x=np.arange(1,6)
y=[10, 15, 12, 18, 14]
y_err = [1,2,1.5,2.5,1]

plt.errorbar(x,y,yerr=y_err, fmt='o',ecolor='red',capsize=5)
plt.title("Error bar")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
#Step Plot

x=np.arange(0,10,1)
y=np.random.randint(1,10,10)

plt.step(x,y,where="mid")
plt.scatter(x,y,color="red")
plt.title("Step Plot")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
# Event occurance (spikes) at certain points

#Stem plot

x=np.linspace(0,2*np.pi,10)
y=np.sin(x)
# plt.plot(x,y)
plt.stem(x,y,basefmt="gray",linefmt="blue",markerfmt="red")
plt.title("stem plot")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

# Random 3D data

x=np.random.rand(50)
y=np.random.rand(50)
z=np.random.rand(50)
fig=plt.figure()
ax=fig.add_subplot(111,projection="3d")
ax.scatter(x,y,z)
plt.show()
```

```python
# contour plot

x=np.linspace(-3,3,100)
y=np.linspace(-3,3,100)
x,y=np.meshgrid(x,y)
z=np.sin(x**2+y**2)

plt.contour(x,y,z,levels=15,cmap="viridis")
plt.colorbar()
plt.show()
```

```python
#Bubble Plot

x=np.random.rand(30)
y=np.random.rand(30)
sizes=np.random.randint(100,1000,30)
plt.scatter(x,y,s=sizes,c=sizes,alpha=0.5,cmap="viridis")
plt.title("Bubble Plot")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
