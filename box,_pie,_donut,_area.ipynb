```python
#Box Plot
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(30)
data1=np.random.normal(50,10,200) #mean=50, #std=10
data2=np.random.normal(60,15,200)
data3=np.random.normal(70,20,200)

plt.boxplot([data1,data2,data3],label=["Group1","Group2","Group3"])
plt.title("Box Plot")
plt.xlabel("Data")
plt.ylabel("Value")
plt.show()
```

```python
#Box Plot
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(30)
data1=np.random.normal(50,10,200) #mean=50, #std=10
data2=np.random.normal(60,15,200)

plt.boxplot([data1,data2],label=["Group1","Group2"],patch_artist=True,boxprops=dict(facecolor="lightblue"),medianprops=dict(color="red",linewidth=2),whiskerprops=dict(color="green",linewidth=2),capprops=dict(color="orange",linewidth=2))
plt.title("Box Plot")
plt.xlabel("Data")
plt.ylabel("Value")
plt.show()
```

```python
#Pie Chart

size=[30,20,25,25]
label=["A","B","C","D"]
colors=["skyblue","lightgreen","orange","gold"]

plt.pie(size,labels=label,colors=colors,autopct='%1.1f%%',startangle=90)

#autopct = automatic percentage
#startangle= it represnts the angle where the first slice of the chart begins
plt.title("Pie Chart")
plt.show()
```

```python
#Donut Chart

plt.pie(size,labels=label,wedgeprops=dict(width=0.5))
plt.title("Donut Chart")
plt.show()
```

```python
#Area Plot

x=np.linspace(0,10,100)
y1=np.sin(x)
y2=np.cos(x)

plt.fill_between(x,y1,color="skyblue",alpha=0.5,label="sin(x)")
plt.fill_between(x,y2,color="salmon",alpha=0.5,label="cos(x)")

plt.title("Area Plot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.legend()
plt.show()
```
