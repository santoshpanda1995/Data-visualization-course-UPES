```python
import matplotlib.pyplot as plt
plt.plot(df["month"],df["sales"])
plt.title("Monthly sales distribution")
plt.xlabel("month")
plt.ylabel("sales")
plt.show()
```

```python
plt.plot(df["month"],df["sales"],color="blue",label="sales",marker="o")
plt.plot(df["month"],df["profit"],color="red",label="profit",marker="s",markerfacecolor="blue",markeredgecolor="green")
plt.title("Monthly sales distribution")
plt.xlabel("month")
plt.ylabel("values")
plt.legend()
plt.show()
```

```python
#Bar Graph

categories=["A","B","C","D"]
Values=[3,5,7,9]
plt.bar(categories,Values,color="orange",width=0.4,edgecolor="black")
plt.title("Bar Graph")
plt.xlabel("Categories")
plt.ylabel("values")
plt.show()
```

```python
#Horizontal Bar Graph

categories=["A","B","C","D"]
Values=[3,5,7,9]
plt.barh(categories,Values,color="orange",edgecolor="black")
plt.title("Bar Graph")
plt.xlabel("Categories")
plt.ylabel("values")
plt.show()
```

```python
#Annotate Bar Graph

categories=["A","B","C","D"]
Values=[3,5,7,9]
plt.bar(categories,Values,color="orange",width=0.4,edgecolor="black")

#Annotations
Labels=["very low","low","medium","high"]
for i,label in enumerate(Labels):
  plt.text(i,Values[i]+0.3,label,ha="center")

plt.annotate("Medium Bar",xy=("C",7),xytext=("C",8.5),arrowprops={"facecolor":"black"})

plt.title("Bar Graph")
plt.xlabel("Categories")
plt.ylabel("values")
plt.show()
```

```python
from matplotlib import color_sequences
import numpy as np
#Multiple Bar Graphs

categories=["A","B","C","D"]
Values1=[3,5,7,9]
Values2=[4,6,7,5]

x=np.arange(len(categories))
width=0.35

plt.bar(x-width/2,Values1,width,label="2024")
plt.bar(x+width/2,Values2, width,label="2026")
plt.plot(Values1,color="red")

plt.xticks(x,categories)
plt.title("Bar Graph")
plt.xlabel("Categories")
plt.ylabel("values")
plt.legend()
plt.show()
```
